# Revision APRX modular

Notebook para revisar mapas, layers, fuentes de datos y attachments usando funciones reutilizables desde `core.aprx_inspector`.

```python
from pathlib import Path

from core.aprx_inspector import inspect_aprx, list_aprx_layers, list_aprx_maps

aprx_path = r"\\amssclgis08.ams.gmams.cl\CL_MLP_PAO\01_Proyectos_ArcGIS\APRX\VISOR TERRITORIAL SIG PAO OFFLINE.aprx"
project_dir = Path.cwd()
project_dir
```

## 1. Listar mapas

```python
maps_df = list_aprx_maps(aprx_path)
maps_df
```

## 2. Seleccionar mapa

```python
# Opcion A: seleccionar por indice.
selected_map_index = 0

# Opcion B: seleccionar por nombre exacto. Deja None para usar el indice.
selected_map_name = None

selected_map_name
```

## 3. Inspeccionar layers del mapa

```python
inspection = inspect_aprx(
    aprx_path=aprx_path,
    map_index=selected_map_index,
    map_name=selected_map_name,
    include_connection_info=True,
)

layers_df = inspection["layers"]
layers_with_sources_df = inspection["layers_with_sources"]
layers_with_attachments_df = inspection["layers_with_attachments"]

layers_df.head()
```

## 4. Layers con fuentes de datos

```python
layers_with_sources_df[[
    "order",
    "depth",
    "group_path",
    "layer_path",
    "is_broken",
    "has_attachments",
    "data_source",
]]
```

## 5. Layers con attachments

```python
layers_with_attachments_df[[
    "order",
    "depth",
    "group_path",
    "layer_path",
    "layer_name",
    "is_broken",
    "data_source",
]]
```

## 6. Exportar resultados

```python
def sanitize_filename(value):
    return "".join(char if char.isalnum() or char in "-_" else "_" for char in value).strip("_")


selected_map_label = selected_map_name or maps_df.loc[selected_map_index, "map_name"]
safe_map_name = sanitize_filename(selected_map_label)

layers_csv = project_dir / f"layers_{safe_map_name}.csv"
sources_csv = project_dir / f"layers_with_sources_{safe_map_name}.csv"
attachments_csv = project_dir / f"layers_with_attachments_{safe_map_name}.csv"

layers_df.to_csv(layers_csv, index=False, encoding="utf-8-sig")
layers_with_sources_df.to_csv(sources_csv, index=False, encoding="utf-8-sig")
layers_with_attachments_df.to_csv(attachments_csv, index=False, encoding="utf-8-sig")

layers_csv, sources_csv, attachments_csv
```
